```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# 3.3. Load dataset
train_df = pd.read_csv("train.csv")
test_df = pd.read_csv("test.csv")

def load_images(df):
    pixels = np.vstack(df["pixels"].str.split().map(
        lambda values: np.asarray(values, dtype=np.uint8)
    ))
    return pixels.reshape(-1, 48, 48, 1)

x_all = load_images(train_df)
x_test = load_images(test_df)

# Chia dữ liệu train thành train/validation; seed cố định để kết quả có thể lặp lại
x_train, x_valid, y_train, y_valid = train_test_split(
    x_all,
    train_df["emotion"].to_numpy(),
    test_size=0.2,
    random_state=42,
    stratify=train_df["emotion"]
)

# 3.4. Normalize dataset
label_encoder = LabelEncoder()
label_encoder.fit(train_df["emotion"])

y_train = np.eye(len(label_encoder.classes_))[label_encoder.transform(y_train)]
y_valid = np.eye(len(label_encoder.classes_))[label_encoder.transform(y_valid)]

# Giống 0.0-1.0 normalization trong mô hình học sâu
x_train = x_train.astype("float32") / 255.0
x_valid = x_valid.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("Classes:", label_encoder.classes_)
print("x_train:", x_train.shape, "y_train:", y_train.shape)
print("x_valid:", x_valid.shape, "y_valid:", y_valid.shape)
print("x_test:", x_test.shape)
print("Pixel range:", x_train.min(), x_train.max())
```
